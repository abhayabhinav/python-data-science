```python
def add_num(x, y , z=None):
    if z != None:
        sum = x + y + z
    else:
        sum = x + y
    return sum
```

# problem - Get avg mpg by cylinder in cty
1. Get unique cylinder
2. Get lenght of each type of unique cylinder
3. Get Sum of cty where cylinder is x
4. Find avg for each of them

```python
cylinders = set(d['cyl'] for d in mpg)
cylinders
```

```python
cty_cylinder_mpg = []
for c in cylinders:
    sum = 0
    cylinder_count = 0
    for d in mpg:
        if d['cyl'] == c:
            sum+= float(d['cty'])
            cylinder_count+= 1
    cty_cylinder_mpg.append((c, sum/cylinder_count))
cty_cylinder_mpg.sort(key = lambda x:x[0])
cty_cylinder_mpg
```

# Get the car size list , avg mpg in city by size of the car

```python
car_class = set(d['class'] for d in mpg)
car_class
```

```python
type(car_class)
```

```python
car_mpg_by_class = []
for c in car_class:
    sum = 0
    count = 0
    avg_by_class = 0
    for d in mpg:
        if d['class'] == c:
            sum+= float(d['cty'])
            count+= 1
    avg_by_class = sum/count
    car_mpg_by_class.append((c, avg_by_class))
# lambda argument : what_to_return
car_mpg_by_class.sort(key=lambda x : x[1])
car_mpg_by_class
```
